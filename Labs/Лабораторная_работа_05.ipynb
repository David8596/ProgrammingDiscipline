Определите количество пар, в которых оба элемента больше, чем среднее арифметическое всех чисел в файле, а их сумма оканчивается на 31. В ответе запишите два числа: сначала количество найденных пар, а затем – максимальную сумму элементов таких пар. ## Решение

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

```python
# Создаем пустой список
data = []
# Выполняем операции чтения и заполнение списка
with open('data/data_5/pr5.txt') as F:
  for line in F:
    digit = int(line)
    data.append(digit)

# Посмотрим, что мы считали...
print(len(data))
print(data[:10])
```

Теперь ответим на поставленные вопросы использую список `data`

```python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)
```

```python
# Подсчет количества подходящих пар
counter = 0
max_sum = -10000
for i in range(len(data)-1):
  # пара элементов
  x, y = data[i], data[i+1]
  # проверяем, подходит ли пара
  if x>sr and y>sr and str(x+y)[-2:] == '31':
    counter +=1
    max_sum = max(max_sum, x+y)
    # print(x, y, x+y)

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
```

Задание варианта 9: В  файле  pr5_1.txt  содержится  последовательность  целых  чисел. 
Элементы последовательности могут принимать целые значения от 
0 до 10 000 включительно. Определите количество пар, в которых 
хотя  бы  один  из  двух  элементов  меньше,  чем  среднее 
арифметическое  всех  чисел  в  файле,  и  хотя  бы  один  из  двух 
элементов делится на 7 и не делится на 3, 11 и 13. В ответе запишите 
два  числа:  сначала  количество  найденных  пар,  а  затем  – минимальную  сумму  элементов  таких  пар.  В  данной  задаче  под 
парой  подразумевается  два  идущих  подряд  элемента 
последовательности.

```python
a = [int(i) for i in open('data/data_5/pr5.txt')]
sr_ar = sum(a)/len(a)

res = []

def id_del(p,o):
    return abs(p)%o==0

for x1,x2 in zip(a, a[1:]):
    if (x1<sr_ar or x2<sr_ar) and (id_del(x1,7) or id_del(x2,7)) and ((not id_del(x1,3)) or (not id_del(x2,3))) and ((not id_del(x1,11)) or (not id_del(x2,11))) and ((not id_del(x1,13)) or (not id_del(x2,13))):
        res.append(x1+x2)
print(f'Количество подходящих пар = {len(res)}')
print(f'Минимальная сумма = {min(res)}')
```
