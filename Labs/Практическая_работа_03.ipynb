# Лабораторная работа 3. Пользовательские функции

## Задание

Напишите функцию, которая по трем переданным ей числам
определяет могут ли эти числа быть сторонами одного и того же
треугольника. Функция возвращает True или False.

## Решение

### Объявление функции

```python
def is_tr(a, b, c):
  return a+b>c and b+c>a and a+c>b
```

### Демонстрация использования функции

```python
x, y, z = 3, 4, 5
if is_tr(x, y, z):
  print(f'Треугольник со сторонами {x}, {y}, {z} существует')
else:
  print(f'Треугольник со сторонами {x}, {y}, {z} не существует')

x, y, z = 1, 4, 10
if is_tr(x, y, z):
  print(f'Треугольник со сторонами {x}, {y}, {z} существует')
else:
  print(f'Треугольник со сторонами {x}, {y}, {z} не существует')
```
