9. == - сравнение на равенство значений
!= - неравенство значений
<,> - меньше, больше
<=, >= - меньше либо равно, больше либо равно
and - логическая И
or - логическое ИЛИ
not - меняет False на True и наоборот

```python
res1 = 5 + 3 > 10 or not 4 == 4
# Сначала сложение, сравнение 8 > 10 - False
# 4 == 4 - True, not True - False
# False or False - False
print(res1)

res2 = 10 != 5 and 3 <= 3
# 10 != 5 - True
# 3 <= 3 - True
# True and True - True
print(res2)

res3 = not (5 >= 2) or 10 < 20 and 7 == 7
# 5 >= 2 - True, not True - False
# 10 < 20 - True
# False or True - True
# True and 7==7(True) - True
print(res3)
```

14. При попытке изменения неизменяемых данных всегда создаётся новый объект изменённой версии заместо старого в памяти. 

```python
age = 17
print(id(age))
age += 1
print(id(age))
```

```python
list1 = [1, 2, 3]
print(id(list1))
list1.append(6)
print(id(list1))
```

т.е. при создании массива данных, в памяти сохранились значения массива и при изменении самого массива через переменную, мы ссылаемся на ту же область в памяти, меняя содержимое => массив является изменяемым типом данным, а таже ссылочным типом
